### 2. Problem Definition
Public bicycle-sharing schemes enhance urban mobility, reduce automotive traffic congestion, and lower carbon emissions. Maintaining operational efficiency requires matching bicycle availability at docking stations with fluctuating spatial and temporal demand. Severe shortages frustrate commuters, while oversupply causes station congestion and unnecessary logistics costs for rebalancing.

**Analytical Objective**:
Formulate a supervised regression task to predict the continuous hourly quantity of rented bicycles ($y = \text{Rented Bike Count}$) based on meteorological conditions (temperature, humidity, solar radiation, precipitation, visibility, wind speed) and calendar indicators (hour, season, day of week, rush hour, holidays, functioning status).

**Mathematical Formulation**:
Given feature vector $\mathbf{x}_i \in \mathbb{R}^d$, learn a predictive function $f: \mathbb{R}^d \to \mathbb{R}$ minimizing expected test loss (evaluated via $R^2$, RMSE, and MAE):
$$\min_f \frac{1}{N} \sum_{i=1}^N \mathcal{L}(y_i, f(\mathbf{x}_i))$$

### 6. Data Cleaning & Outlier Audit
We formalize data cleaning findings and perform an outlier audit across numerical features and the target variable using the Interquartile Range (IQR) method.

```python
# Explicitly verify missing values and duplicate rows
missing_counts = df.isnull().sum()
duplicate_rows = df.duplicated().sum()

print("--- DATA INTEGRITY VERIFICATION ---")
print(f"Total missing values across all columns: {missing_counts.sum()}")
print(f"Total duplicate rows in dataset: {duplicate_rows}")

# Outlier analysis using the Interquartile Range (IQR) method
iqr_records = []
for col in numerical_cols + ['Rented_Bike_Count']:
    q25 = df[col].quantile(0.25)
    q75 = df[col].quantile(0.75)
    iqr = q75 - q25
    lower_bound = q25 - 1.5 * iqr
    upper_bound = q75 + 1.5 * iqr
    outlier_count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    pct_outliers = (outlier_count / len(df)) * 100
    iqr_records.append({
        'Feature': col,
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2),
        'Outlier Count': outlier_count,
        'Outlier %': round(pct_outliers, 2)
    })

outlier_df = pd.DataFrame(iqr_records)
display(outlier_df)
```

### 7. Train/Test Split (with Quantile Stratification)
To evaluate models without data leakage, we split the dataset into an **80% training set (7,008 samples)** and a **held-out 20% test set (1,752 samples)**.

**Quantile Stratification Methodology**:
Because `Rented_Bike_Count` is a continuous, right-skewed variable, standard random splitting risks an unbalanced representation of rare peak-demand observations between sets. We partition the continuous target into 10 quantile bins solely to stratify the split via `train_test_split(..., stratify=y_bins, random_state=42)`.

> [!IMPORTANT]
> The quantile bin column is discarded immediately after partitioning and is never supplied to any model or preprocessor, preventing target leakage.

**Important Time-Series Limitation**:
> [!NOTE]
> Because the course guidelines specify a random train/test split, this evaluation measures interpolation on held-out samples drawn from the same recorded annual period rather than true forward-in-time forecasting. Temporal generalization should be evaluated separately in future production work using a chronological split.

```python
from sklearn.model_selection import train_test_split

target_col = 'Rented_Bike_Count'
y = df[target_col].copy()
X_raw = df.drop(columns=[target_col]).copy()

# Generate temporary quantile bins for stratified splitting
y_strat_bins = pd.qcut(y, q=10, duplicates='drop')

# Execute stratified 80:20 split
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y_strat_bins
)

print(f"Training feature set shape: {X_train_raw.shape} | Training labels: {y_train.shape}")
print(f"Test feature set shape:     {X_test_raw.shape} | Test labels:     {y_test.shape}")

# Verify distributional similarity across train and test sets
split_check = pd.DataFrame({
    'Metric': ['Mean', 'Std Dev', 'Median', 'Min', 'Max'],
    'Full Dataset': [y.mean(), y.std(), y.median(), y.min(), y.max()],
    'Train Set': [y_train.mean(), y_train.std(), y_train.median(), y_train.min(), y_train.max()],
    'Test Set': [y_test.mean(), y_test.std(), y_test.median(), y_test.min(), y_test.max()]
}).round(2)
display(split_check)
```

### 8. Feature Engineering
We introduce domain-motivated temporal features derived from `Date` and `Hour`:
1. **Calendar Date Components**:
   - `Month` (1–12): Represents seasonal progression beyond the 4 broad season categories.
   - `Day` (1–31): Captures intra-month dynamics.
   - `DayOfWeek` (0–6): Represents day-to-day commute patterns.
   - `Is_Weekend` (0 or 1): Binary indicator (Saturday=5, Sunday=6) reflecting leisure vs. commuter profiles.
2. **Commuter Rush Hour Indicator (`Rush_Hour`)**:
   - Commuter bike traffic peaks during morning (08:00–09:59) and evening (18:00–20:59) hours on working weekdays (`Is_Weekend == 0`).
3. **Cyclical Hour Representations (`Hour_sin`, `Hour_cos`)**:
   - Hour-of-day is naturally circular (hour 23 is immediately adjacent to hour 0). Applying sine and cosine harmonic transformations preserves this continuous circular topology:
     $$\text{Hour}_{\sin} = \sin\left(\frac{2\pi \times \text{Hour}}{24}\right), \quad \text{Hour}_{\cos} = \cos\left(\frac{2\pi \times \text{Hour}}{24}\right)$$

**Leakage Prevention**:
All transformations rely purely on the timestamp of each individual observation; no aggregations across the dataset or target values are computed.

```python
def engineer_features(data):
    df_out = data.copy()
    
    if not pd.api.types.is_datetime64_any_dtype(df_out['Date']):
        dt_series = pd.to_datetime(df_out['Date'], format='%d/%m/%Y')
    else:
        dt_series = df_out['Date']
        
    df_out['Month'] = dt_series.dt.month
    df_out['Day'] = dt_series.dt.day
    df_out['DayOfWeek'] = dt_series.dt.dayofweek
    df_out['Is_Weekend'] = (df_out['DayOfWeek'] >= 5).astype(int)
    
    # Cyclical hour encoding
    df_out['Hour_sin'] = np.sin(2 * np.pi * df_out['Hour'] / 24.0)
    df_out['Hour_cos'] = np.cos(2 * np.pi * df_out['Hour'] / 24.0)
    
    # Commuter Rush Hour indicator (Weekdays: 8-9 AM, 6-8 PM)
    df_out['Rush_Hour'] = (
        ((df_out['Hour'].isin([8, 9])) | (df_out['Hour'].isin([18, 19, 20]))) & 
        (df_out['Is_Weekend'] == 0)
    ).astype(int)
    
    # Drop raw date column and temporary datetime column
    cols_to_drop = [c for c in ['Date', 'Date_dt'] if c in df_out.columns]
    df_out = df_out.drop(columns=cols_to_drop)
    
    return df_out

# Apply feature engineering to train and test sets independently
X_train_fe = engineer_features(X_train_raw)
X_test_fe = engineer_features(X_test_raw)

print(f"Feature engineering complete. Columns: {X_train_fe.columns.tolist()}")
print(f"Engineered training shape: {X_train_fe.shape} | Test shape: {X_test_fe.shape}")
X_train_fe.head(3)
```

### 9. Encoding and Preprocessing Pipeline
We build a scikit-learn `ColumnTransformer`:
- **Categorical Columns** (`Seasons`, `Holiday`, `Functioning_Day`): Processed via `OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False)` to prevent dummy variable traps and output dense arrays.
- **Numerical Columns**: Standardized to zero mean and unit variance using `StandardScaler()`.

> [!CAUTION]
> **Strict Leakage Prevention**:
> `preprocessor.fit()` is invoked **strictly on `X_train_fe`**. The fitted transformer is then used to `transform()` both `X_train_fe` and `X_test_fe`.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

categorical_features = ['Seasons', 'Holiday', 'Functioning_Day']
numerical_features = [c for c in X_train_fe.columns if c not in categorical_features]

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False), categorical_features)
    ]
)

# Fit strictly on training data
X_train_prep = preprocessor.fit_transform(X_train_fe)
X_test_prep = preprocessor.transform(X_test_fe)

# Retrieve transformed column names
cat_encoder = preprocessor.named_transformers_['cat']
cat_feature_names = cat_encoder.get_feature_names_out(categorical_features).tolist()
all_feature_names = numerical_features + cat_feature_names

print(f"Preprocessed feature count: {X_train_prep.shape[1]}")
print("Feature names after preprocessing:", all_feature_names)
```

### 10. Model Training (All 10 Required Algorithms)
We implement, train, and benchmark all ten required regression algorithms on the exact same training set (`X_train_prep`, `y_train`) and evaluate on the exact same held-out test set (`X_test_prep`, `y_test`):
1. **Linear Regression** (Baseline model; coefficient interpretation)
2. **Ridge Regression** ($L_2$ regularization)
3. **Lasso Regression** ($L_1$ regularization; feature sparsity)
4. **ElasticNet Regression** (Combined $L_1 + L_2$ regularization)
5. **Polynomial Regression** (Controlled Degree 2 interaction pipeline)
6. **Decision Tree Regressor** (Non-linear recursive partitioning)
7. **Random Forest Regressor** (Ensemble bagging)
8. **Gradient Boosting Regressor** (Ensemble sequential boosting)
9. **Support Vector Regressor (SVR)** (Kernel-based regression on standardized features)
10. **K-Nearest Neighbors Regressor** (Instance-based distance weighting)

```python
from sklearn.metrics import r2_score, root_mean_squared_error, mean_absolute_error
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor

baseline_models = {}
test_predictions = {}
baseline_metrics = []

def evaluate_baseline_model(name, model, X_tr, y_tr, X_te, y_te):
    model.fit(X_tr, y_tr)
    preds = model.predict(X_te)
    r2 = r2_score(y_te, preds)
    rmse = root_mean_squared_error(y_te, preds)
    mae = mean_absolute_error(y_te, preds)
    
    baseline_models[name] = model
    test_predictions[name] = preds
    baseline_metrics.append({
        'Model': name,
        'R2': r2,
        'RMSE': rmse,
        'MAE': mae
    })
    print(f"-> Trained {name:30s} | R2: {r2:.4f} | RMSE: {rmse:.2f} | MAE: {mae:.2f}")

print("--- TRAINING 10 BASELINE REGRESSION MODELS ---")

# 1. Linear Regression
evaluate_baseline_model('Linear Regression', LinearRegression(), X_train_prep, y_train, X_test_prep, y_test)

# 2. Ridge Regression
evaluate_baseline_model('Ridge Regression', Ridge(alpha=1.0, random_state=RANDOM_STATE), X_train_prep, y_train, X_test_prep, y_test)

# 3. Lasso Regression
evaluate_baseline_model('Lasso Regression', Lasso(alpha=1.0, random_state=RANDOM_STATE), X_train_prep, y_train, X_test_prep, y_test)

# 4. ElasticNet Regression
evaluate_baseline_model('ElasticNet Regression', ElasticNet(alpha=1.0, l1_ratio=0.5, random_state=RANDOM_STATE), X_train_prep, y_train, X_test_prep, y_test)

# 5. Polynomial Regression (Degree 2 controlled pipeline)
poly_pipeline = Pipeline([
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('lr', LinearRegression())
])
evaluate_baseline_model('Polynomial Regression (deg 2)', poly_pipeline, X_train_prep, y_train, X_test_prep, y_test)

# 6. Decision Tree Regressor
evaluate_baseline_model('Decision Tree Regressor', DecisionTreeRegressor(random_state=RANDOM_STATE), X_train_prep, y_train, X_test_prep, y_test)

# 7. Random Forest Regressor
evaluate_baseline_model('Random Forest Regressor', RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1), X_train_prep, y_train, X_test_prep, y_test)

# 8. Gradient Boosting Regressor
evaluate_baseline_model('Gradient Boosting Regressor', GradientBoostingRegressor(random_state=RANDOM_STATE), X_train_prep, y_train, X_test_prep, y_test)

# 9. Support Vector Regressor (SVR baseline)
evaluate_baseline_model('Support Vector Regressor', SVR(C=1.0, epsilon=0.1), X_train_prep, y_train, X_test_prep, y_test)

# 10. K-Nearest Neighbors Regressor
evaluate_baseline_model('K-Nearest Neighbors Regressor', KNeighborsRegressor(n_neighbors=5, n_jobs=-1), X_train_prep, y_train, X_test_prep, y_test)
```

### 12. Controlled Polynomial Degree Comparison
The capstone guidelines specify comparing polynomial degrees. We implement a controlled empirical comparison evaluating Degree 1 (standard linear baseline) and Degree 2 (quadratic plus interaction features) on the same train/test split.

```python
# Empirical comparison of Polynomial Degrees
poly_degrees = [1, 2]
poly_comparison_records = []

for deg in poly_degrees:
    if deg == 1:
        pipe = Pipeline([
            ('lr', LinearRegression())
        ])
        f_count = X_train_prep.shape[1]
    else:
        pipe = Pipeline([
            ('poly', PolynomialFeatures(degree=deg, include_bias=False)),
            ('lr', LinearRegression())
        ])
        f_count = pipe.named_steps['poly'].fit_transform(X_train_prep[:5]).shape[1]
        
    pipe.fit(X_train_prep, y_train)
    p_preds = pipe.predict(X_test_prep)
    
    poly_comparison_records.append({
        'Polynomial Degree': f'Degree {deg}',
        'Feature Count': f_count,
        'Test R2': round(r2_score(y_test, p_preds), 4),
        'Test RMSE': round(root_mean_squared_error(y_test, p_preds), 2),
        'Test MAE': round(mean_absolute_error(y_test, p_preds), 2),
        'Computational Complexity': 'O(d) - Fast' if deg == 1 else 'O(d^2) - Manageable'
    })

# Add theoretical analysis row for Degree 3
poly_deg3_test = PolynomialFeatures(degree=3, include_bias=False)
deg3_count = poly_deg3_test.fit_transform(X_train_prep[:2]).shape[1]
poly_comparison_records.append({
    'Polynomial Degree': 'Degree 3 (Omitted)',
    'Feature Count': deg3_count,
    'Test R2': np.nan,
    'Test RMSE': np.nan,
    'Test MAE': np.nan,
    'Computational Complexity': 'O(d^3) - Memory/Overfitting Risk'
})

poly_comparison_df = pd.DataFrame(poly_comparison_records)
print("=== POLYNOMIAL DEGREE COMPARISON TABLE ===")
display(poly_comparison_df)
```

**Observation & Complexity Assessment**:
- **Degree 1 (Linear)** uses the 21 preprocessed features, achieving $R^2 = 0.6191$ and $\text{RMSE} = 397.72$.
- **Degree 2** expands the feature set to **252 interaction and quadratic features**, improving test $R^2$ to **0.8450** and lowering RMSE to **253.71**.
- **Degree 3** would expand the feature matrix to **2,024 features** for 7,008 training samples. Fitting unregularized linear regression on 2,024 highly collinear polynomial features carries severe risks of memory bloat, matrix ill-conditioning, and run-time instability. Degree 2 is therefore selected as the balanced, computationally tractable polynomial representation.

### 13. Hyperparameter Tuning (Leak-Free Pipeline Architecture)
As required by Section C3 of the rubric, we apply `GridSearchCV` to **two selected models**:
1. **Support Vector Regressor (SVR)**: Tuning $C$ and $\epsilon$ to address the scale mismatch identified in baseline testing.
2. **Random Forest Regressor**: Tuning `n_estimators`, `max_depth`, and `min_samples_split` to optimize ensemble capacity.

> [!IMPORTANT]
> **Strict Leakage Prevention via Pipeline**:
> Preprocessing and the regression estimator are wrapped in a single `sklearn.pipeline.Pipeline`. Raw engineered features `X_train_fe` and `y_train` are passed directly into `GridSearchCV`. Inside each internal CV fold, preprocessing transformations are fitted strictly on that fold's training slice, completely preventing data leakage into validation folds.

```python
from sklearn.model_selection import GridSearchCV

print("--- HYPERPARAMETER TUNING VIA LEAK-FREE PIPELINES ---")

# 1. Support Vector Regressor Tuning Pipeline
print("1. Tuning Support Vector Regressor (SVR)...")
svr_pipe = Pipeline([
    ('prep', preprocessor),
    ('svr', SVR())
])
svr_param_grid = {
    'svr__C': [100, 500, 1000],
    'svr__epsilon': [5.0, 10.0, 20.0]
}
svr_grid = GridSearchCV(
    estimator=svr_pipe,
    param_grid=svr_param_grid,
    cv=3,
    scoring='r2',
    n_jobs=-1,
    verbose=0
)
svr_grid.fit(X_train_fe, y_train)
best_svr_pipe = svr_grid.best_estimator_
svr_tuned_preds = best_svr_pipe.predict(X_test_fe)

svr_base_row = master_comparison_df.loc[master_comparison_df['Model'] == 'Support Vector Regressor'].iloc[0]
svr_tuned_r2 = r2_score(y_test, svr_tuned_preds)
svr_tuned_rmse = root_mean_squared_error(y_test, svr_tuned_preds)
svr_tuned_mae = mean_absolute_error(y_test, svr_tuned_preds)

print(f"   Best SVR Parameters: {svr_grid.best_params_}")
print(f"   SVR Baseline R2: {svr_base_row['R2']:.4f} -> Tuned R2: {svr_tuned_r2:.4f} (Delta: +{svr_tuned_r2 - svr_base_row['R2']:.4f})")

# 2. Random Forest Regressor Tuning Pipeline
print("2. Tuning Random Forest Regressor...")
rf_pipe = Pipeline([
    ('prep', preprocessor),
    ('rf', RandomForestRegressor(random_state=RANDOM_STATE))
])
rf_param_grid = {
    'rf__n_estimators': [100, 200],
    'rf__max_depth': [20, None],
    'rf__min_samples_split': [2, 5]
}
rf_grid = GridSearchCV(
    estimator=rf_pipe,
    param_grid=rf_param_grid,
    cv=3,
    scoring='r2',
    n_jobs=-1,
    verbose=0
)
rf_grid.fit(X_train_fe, y_train)
best_rf_pipe = rf_grid.best_estimator_
rf_tuned_preds = best_rf_pipe.predict(X_test_fe)

rf_base_row = master_comparison_df.loc[master_comparison_df['Model'] == 'Random Forest Regressor'].iloc[0]
rf_tuned_r2 = r2_score(y_test, rf_tuned_preds)
rf_tuned_rmse = root_mean_squared_error(y_test, rf_tuned_preds)
rf_tuned_mae = mean_absolute_error(y_test, rf_tuned_preds)

print(f"   Best RF Parameters: {rf_grid.best_params_}")
print(f"   RF Baseline R2: {rf_base_row['R2']:.4f} -> Tuned R2: {rf_tuned_r2:.4f} (Delta: +{rf_tuned_r2 - rf_base_row['R2']:.4f})")
```

```python
# Dedicated Hyperparameter Tuning Results Table
tuning_summary_df = pd.DataFrame([
    {
        'Model': 'Support Vector Regressor',
        'Best Hyperparameters': str(svr_grid.best_params_),
        'Baseline R2': f"{svr_base_row['R2']:.4f}",
        'Tuned R2': f"{svr_tuned_r2:.4f}",
        'Baseline RMSE': f"{svr_base_row['RMSE']:.2f}",
        'Tuned RMSE': f"{svr_tuned_rmse:.2f}",
        'Baseline MAE': f"{svr_base_row['MAE']:.2f}",
        'Tuned MAE': f"{svr_tuned_mae:.2f}",
        'Delta R2': f"+{svr_tuned_r2 - svr_base_row['R2']:.4f}",
        'Delta RMSE': f"-{svr_base_row['RMSE'] - svr_tuned_rmse:.2f}",
        'Delta MAE': f"-{svr_base_row['MAE'] - svr_tuned_mae:.2f}"
    },
    {
        'Model': 'Random Forest Regressor',
        'Best Hyperparameters': str(rf_grid.best_params_),
        'Baseline R2': f"{rf_base_row['R2']:.4f}",
        'Tuned R2': f"{rf_tuned_r2:.4f}",
        'Baseline RMSE': f"{rf_base_row['RMSE']:.2f}",
        'Tuned RMSE': f"{rf_tuned_rmse:.2f}",
        'Baseline MAE': f"{rf_base_row['MAE']:.2f}",
        'Tuned MAE': f"{rf_tuned_mae:.2f}",
        'Delta R2': f"+{rf_tuned_r2 - rf_base_row['R2']:.4f}",
        'Delta RMSE': f"-{rf_base_row['RMSE'] - rf_tuned_rmse:.2f}",
        'Delta MAE': f"-{rf_base_row['MAE'] - rf_tuned_mae:.2f}"
    }
])

print("=== DEDICATED HYPERPARAMETER TUNING RESULTS TABLE ===")
display(tuning_summary_df)
```

**Observation on Tuning Impact**:
- **SVR**: Under default $C=1.0$ and $\epsilon=0.1$, SVR underperformed ($R^2 = 0.2794$). By expanding the cost penalty to $C=1000$ and tolerance to $\epsilon=20.0$, the tuned SVR achieves an $R^2$ of **0.9046**, an improvement of **+0.6252**.
- **Random Forest**: Already achieved strong baseline performance ($R^2 = 0.9429$). Tuning `n_estimators=200` yields $R^2 = 0.9434$, with slight error reductions (RMSE: 154.03 $\to$ 153.25; MAE: 93.60 $\to$ 93.30).
